# Backpropagation: First Code (2 inputs, 2 hidden, 2 outputs)

**Retail example:**
- Input 1 = how much we spend on ads (scaled 0 to 1)
- Input 2 = discount given (scaled 0 to 1)
- Output 1 = sales of Product A (scaled 0 to 1)
- Output 2 = sales of Product B (scaled 0 to 1)

The network learns the weights so that its predictions get close to the real sales.

## Step 1: Import the library

```python
import numpy as np
```

## Step 2: Sigmoid function
Sigmoid squeezes any number into a value between 0 and 1.

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

## Step 3: Our data and starting weights
- `x` = inputs (ads, discount)
- `y` = real answers (sales A, sales B)
- `w1` = weights from input to hidden layer
- `w2` = weights from hidden to output layer
- `b1`, `b2` = biases
- `lr` = learning rate (how big each step is)

```python
x = np.array([0.05, 0.10])   # inputs
y = np.array([0.01, 0.99])   # real sales (targets)

# weights: row = input, column = neuron
w1 = np.array([[0.15, 0.25],
               [0.20, 0.30]])
b1 = np.array([0.35, 0.35])

w2 = np.array([[0.40, 0.50],
               [0.45, 0.55]])
b2 = np.array([0.60, 0.60])

lr = 0.5
```

## Step 4: Forward pass
Move the data from input to output and make a prediction.

```python
# Hidden layer
h_in  = x @ w1 + b1
h_out = sigmoid(h_in)

# Output layer
o_in  = h_out @ w2 + b2
pred  = sigmoid(o_in)

print("Hidden output:", h_out)
print("Prediction   :", pred)
```

## Step 5: Loss (how wrong are we?)
Loss = 1/2 × (prediction − actual)², added for both outputs.

```python
loss = 0.5 * np.sum((pred - y) ** 2)
print("Loss:", loss)
```

## Step 6: Backpropagation
We find how much each weight caused the error, going backward.

- Output layer: `delta_o = (pred - y) * pred * (1 - pred)`
- Hidden layer: `delta_h = (delta_o @ w2.T) * h_out * (1 - h_out)`

```python
# Output layer error
delta_o = (pred - y) * pred * (1 - pred)
grad_w2 = np.outer(h_out, delta_o)
grad_b2 = delta_o

# Hidden layer error
delta_h = (delta_o @ w2.T) * h_out * (1 - h_out)
grad_w1 = np.outer(x, delta_h)
grad_b1 = delta_h

print("Gradient w2:\n", grad_w2)
print("Gradient w1:\n", grad_w1)
```

## Step 7: Update the weights (gradient descent)
new weight = old weight − learning rate × gradient

```python
w2 = w2 - lr * grad_w2
b2 = b2 - lr * grad_b2
w1 = w1 - lr * grad_w1
b1 = b1 - lr * grad_b1

print("New w1:\n", w1)
print("New w2:\n", w2)
```

## Step 8: Train many times
Repeat forward pass, loss, backpropagation, and update. The loss should go down.

```python
for epoch in range(10000):
    # forward
    h_out = sigmoid(x @ w1 + b1)
    pred  = sigmoid(h_out @ w2 + b2)
    loss  = 0.5 * np.sum((pred - y) ** 2)

    # backward
    delta_o = (pred - y) * pred * (1 - pred)
    delta_h = (delta_o @ w2.T) * h_out * (1 - h_out)

    # update
    w2 -= lr * np.outer(h_out, delta_o)
    b2 -= lr * delta_o
    w1 -= lr * np.outer(x, delta_h)
    b1 -= lr * delta_h

    if epoch % 2000 == 0:
        print(f"Epoch {epoch}: loss = {loss:.6f}")

print("\nFinal prediction:", pred)
print("Real answer     :", y)
```
